```python
def draw_M(height):
    for i in range(height):
        for j in range(height + 1):
            # Les conditions pour dessiner le "M"
            if j == 0 or j == height:  # Bord gauche et bord droit
                print("M", end="")
            elif j == i and i <= height // 2:  # Diagonale gauche
                print("M", end="")
            elif j == height - i and i <= height // 2:  # Diagonale droite
                print("M", end="")
            else:
                print(" ", end="")
        print()  # Nouvelle ligne

# Appeler la fonction avec la hauteur désirée
draw_M(7)
```
